# 1. Welcome: Data Science *for* Social Justice

Welcome to DIGHUM 162. This course is built on a simple but unusual premise: **data science is not neutral, and it never has been.** Every dataset is the product of choices — about what to measure, who to count, how to label, what to drop. Every algorithm is the product of choices too. Those choices have consequences, and those consequences fall unevenly across people.

This week, we'll set up some technical foundations and begin orienting ourselves to the course's central question:

> **Who gets counted in data — and who is rendered invisible?**

This question will reappear in every notebook this term. Hold onto it.

## The Four Conceptual Pillars

The course rests on four pillars. We'll touch each one this week and develop them throughout the term:

1. **Anti-colonial + Science and Technology Studies (STS) perspectives.** The dominant story of data science is told from a small set of places, languages, and institutional positions. We try to make those positions visible — and to listen for what they leave unsaid.

2. **Algorithmic bias.** Models trained on biased data reproduce that bias *at scale*. Worse, the math gives the bias a veneer of objectivity. "The algorithm decided" sounds different from "a person decided," but in many cases it shouldn't.

3. **Problematizing data.** Datasets are constructs. Categories are constructs. "Race," "gender," "recidivism," "income," "healthy" — every column in every dataset has a history. Asking how a column got built is not a side question; it *is* the question.

4. **Problematizing data science.** The very phrase "data science" carries assumptions: that knowledge looks like measurement, that good answers come from bigger datasets, that what can't be quantified doesn't matter. We'll keep asking what those assumptions cost.

### Variables and types

A *variable* is just a label that points at a value. Python figures out the type for you.

```python
course_name = "DIGHUM 162"   # a string (text)
week = 1                       # an integer
grade_weight = 0.20            # a float (decimal)
is_first_week = True           # a boolean (True / False)

print(course_name, type(course_name))
print(week, type(week))
print(grade_weight, type(grade_weight))
print(is_first_week, type(is_first_week))
```

### Lists

A list is an ordered collection. Square brackets, items separated by commas.

```python
pillars = ["anti-colonial/STS", "algorithmic bias", "problematizing data", "problematizing data science"]
print(pillars[0])      # Python counts from 0
print(len(pillars))    # how many items
pillars.append("social justice praxis")
pillars
```

### Dictionaries

A dictionary maps *keys* to *values*. Great for labeled data.

```python
grading = {
    "participation": 0.20,
    "weekly_assignments": 0.30,
    "group_project": 0.25,
    "individual_essay": 0.25,
}

print("Participation is worth", grading["participation"] * 100, "percent")
print("Total =", sum(grading.values()))
```

### Loops and conditionals

A `for` loop walks through a collection. An `if` statement runs code only when a condition is true.

```python
for category, weight in grading.items():
    if weight >= 0.25:
        print(category, "is a major component (", weight, ")")
    else:
        print(category, "is a smaller component (", weight, ")")
```

Let's meet a tiny dataset. We'll build it ourselves, in code, so that we can see exactly where it came from. Even this much honesty about a dataset's origin is rare in real-world data science.

But notice: even when I made it up, I had to choose *which columns to include* (name, age, neighborhood, income, prior arrests).
```python
applicants = pd.DataFrame({
    "name":         ["Aisha", "Brian", "Carlos", "Devon", "Esi",  "Frank", "Grace", "Hannah", "Imani", "Jorge"],
    "age":          [22,      28,      35,       19,      41,     33,      27,      45,       24,      38   ],
    "neighborhood":["East",  "West",  "East",   "East",  "West","West",  "East",  "West",   "East",  "West"],
    "income":      [28000,   62000,   31000,    24000,   71000, 58000,   30000,   75000,    27000,   60000],
    "prior_arrests":[0,      0,       1,        2,       0,      0,       0,       0,        1,       0   ],
})
applicants
```

💭 **Reflection**: Look at the columns. What's *missing* that might matter for understanding these ten people? (Race? Gender? Disability? Caregiver status? Citizenship? Whether the prior arrests led to convictions? The reasons for them?)

⚠️ **Warning**: `prior_arrests` looks like a clean number, but think about what it actually measures. *Arrests* are a record of police contact, not of crime. A column that *looks* like it measures "criminality" is in fact measuring something closer to "exposure to law enforcement." We'll see this issue again, in much higher stakes, in Week 2 (COMPAS).

# 5. Reading: *Data Feminism* (D'Ignazio & Klein)

For Monday, you read the introduction to D'Ignazio and Klein's *Data Feminism* (2020). It is the keystone reading of the course. Their seven principles aren't a checklist — they're a way of looking. Here are three to hold onto right now:

**1. Examine power.** Whose perspective shaped this dataset? Who benefits from it being collected? Who could be harmed? Power runs through data, not around it.

**2. Challenge power.** Examining isn't enough. Data work can be used to *contest* unjust arrangements, not just describe them. Think of the work of the COVID Tracking Project, or M.A.P. (Mapping Police Violence), or the Anti-Eviction Mapping Project.

**3. Elevate emotion and embodiment.** The myth of the dispassionate analyst is itself a position. Real people care about what they study; pretending otherwise hides who is being served by the supposed neutrality.

# 6. Reading: *Algorithms of Oppression* (Noble)

For Tuesday, you read Chapter 1 of Safiya Noble's *Algorithms of Oppression* (2018). Noble's central case study: search engines. She documents how Google's autocomplete and image results, when she queried "black girls" in 2010, returned a deluge of pornographic and degrading content — far more than for queries about white girls or Asian girls. The pattern was systematic, replicable, and consequential.

Three takeaways to bring forward:

- **Algorithms are not neutral arbiters.** They reflect the data they're trained on, the choices of their designers, and the commercial incentives of the companies running them.
- **Search is not the same as truth.** Ranking is *not* the same as accuracy or fairness. A first-page result is a result of optimization, not of a vote of the universe.
- **Harm scales.** When millions of people query the same algorithm, individual biases become structural ones.

# 7. A Tiny Bias Demonstration

Let's make Noble's point concrete with a few lines of Python. Imagine we're building a "reliability score" for our `applicants`. Suppose, for the sake of demonstration, the model uses just one feature: neighborhood.

```python
def reliability_score(neighborhood):
    # Suppose someone built this model based on "historical defaults"
    # in their existing data, which happen to skew by neighborhood.
    if neighborhood == "West":
        return 0.85
    else:
        return 0.40

applicants["score"] = applicants["neighborhood"].apply(reliability_score)
applicants[["name", "neighborhood", "income", "score"]]
```

Notice what just happened. The function `reliability_score` doesn't know anyone's *name*. It doesn't know their *income*. It doesn't even know their *individual history*. It just knows the **neighborhood**. And yet, by treating neighborhood as a feature, the function reproduces *whatever pattern was in the data the model-builder used to set those numbers*. If those historical patterns were themselves shaped by redlining, or by uneven access to credit, or by which neighborhoods got policed more — the model now launders that history into a clean-looking decimal.

🔔 **Question**: The function above never mentions race. So is it race-neutral? 

(Hold this question. We'll see in Week 3 that omitting a feature like race is *not* the same as building a race-neutral model — when other features are correlated with race due to structural inequities, those features become *proxies*. The model still learns the disparity; it just hides where it's coming from.)
