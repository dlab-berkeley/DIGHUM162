# 2. The Four Pillars Revisited

Let's revisit the four critical and conceptual pillars from Week 1, this time with concrete examples from the course.

**1. Anti-colonial + STS perspectives.** *Example from Week 4:* the categories "non-Hispanic Black," "Hispanic/Latino," etc., are administrative constructions that have shifted over time. Mapping disparities in these categories surfaces structural injustice — and also reproduces, every time we use them, the racial schema that makes those categories thinkable. Holding both is the work.

**2. Algorithmic bias.** *Example from Week 3:* a logistic regression that never received `race` as a feature still produced racially patterned errors via proxy variables. Bias lives in the data and the world that produced it; choosing a different model doesn't escape it.

**3. Problematizing data.** *Example from Week 2:* `priors_count` looks like an attribute of a person, but is in fact an attribute of *the system that produced the record*. The same act with the same person can yield very different `priors_count` numbers depending on the policing of the place they live.

**4. Problematizing data science.** *Example from Week 5:* Bolukbasi et al.'s "debiasing" paper is a paradigmatic positivist response to bias — a technical fix, scoped to one axis (gender, binary), that Gonen & Goldberg (2019) showed *hides* bias rather than removing it. The very framing of "how do we de-bias the model?" treats the corpus and its world as fixed inputs, when in fact they were the things that produced the harm.

# 4. Choosing a Research Question

A good research question for this course satisfies three criteria:

- **Grounded in a specific dataset.** Not "is the criminal justice system biased?" — too sweeping. Instead: "How does COMPAS's false positive rate change when we equalize across race? What's the human cost of doing so?"
- **Critical, not just descriptive.** Don't stop at "there is a difference between groups X and Y." Ask *why*, what historical and structural patterns produced it, and what political question your finding raises.
- **Small enough to actually finish.** Two weeks is short. A focused, well-executed analysis on one slice of a dataset beats an unfinished moonshot.

### Example RQs you can borrow or modify

- "Do COMPAS error rates differ by gender as well as by race? What do those differences imply about how the criminal-legal system imagines what kind of person is dangerous?"
- "What words distinguish posts about workplace conflict from posts about family conflict in r/AmITheAsshole, and what do those vocabulary differences imply about what each domain is *about* to its participants?"
- "Which US states with the lowest life expectancies also had the highest historical redlining intensities? What does the temporal lag tell us about how slowly geographic injustice unwinds?"
- "Train a community-specific Word2Vec model on a chosen subreddit. Compare its analogies to those of pre-trained GloVe. What does the comparison reveal about how community language differs from generic-corpus language?"
- "For a chosen US city, map life expectancy by census tract alongside historical zoning maps. What is the correlation? What can a map *not* show that interview or oral-history work might?"

# 5. An End-to-End Mini-Pipeline

Let's walk through a complete analysis on a small dataset. Not because the dataset matters — it's a teaching toy — but because the *steps* are the steps your final project will take. The dataset below is synthetic loan-application data (made up by me).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix

rng = np.random.default_rng(42)
n = 200
race = rng.choice(["A", "B"], n, p=[0.5, 0.5])
income = np.where(race == "A",
                  rng.normal(70_000, 15_000, n),
                  rng.normal(55_000, 15_000, n))   # systematic gap
credit_score = np.where(race == "A",
                  rng.normal(720, 60, n),
                  rng.normal(680, 60, n))
years_employed = rng.normal(7, 3, n).clip(0, None)
# Approval depends on income and credit — both shaped by the structural gap
logit = -8 + 0.00008 * income + 0.005 * credit_score + 0.02 * years_employed
prob = 1 / (1 + np.exp(-logit))
approved = (rng.uniform(size=n) < prob).astype(int)

loans = pd.DataFrame({"income": income.round(0), "credit_score": credit_score.round(0),
                      "years_employed": years_employed.round(1),
                      "race": race, "approved": approved})
loans.head()
```

**Step 1: Load and inspect.**

```python
print("Shape:", loans.shape)
print(loans["race"].value_counts())
print("Approval rate overall:", loans["approved"].mean().round(3))
```

**Step 2: Clean — and explicitly note what you're choosing to exclude.**

```python
# Drop rows with missing data — but check first that we're not dropping a particular group disproportionately.
print("Missing per column:")
print(loans.isnull().sum())
loans_clean = loans.dropna()
# If we'd had real data, we'd also note here which categories the data lacks
# (e.g., what races are *not* in the dataset at all — every dataset has erasures).
print("After dropna:", loans_clean.shape)
```

**Step 3: Explore — by group.**

```python
loans_clean.groupby("race").agg({
    "income": "mean", "credit_score": "mean",
    "years_employed": "mean", "approved": "mean"}).round(2)
```

```python
# Visualize approval rates by race
fig, ax = plt.subplots(figsize=(6, 4))
approval_by_race = loans_clean.groupby("race")["approved"].mean()
approval_by_race.plot.bar(ax=ax, color=["steelblue", "tomato"])
ax.set_ylabel("Approval rate")
ax.set_title("Loan approval rate by race (synthetic teaching data)")
ax.set_ylim(0, 1)
for i, v in enumerate(approval_by_race):
    ax.text(i, v + 0.02, f"{v:.2f}", ha="center")
plt.tight_layout()
plt.show()
```

**Step 4: Model — *not using race as a feature*.**

```python
X = loans_clean[["income", "credit_score", "years_employed"]]
y = loans_clean["approved"]
race_col = loans_clean["race"]

X_train, X_test, y_train, y_test, race_train, race_test = train_test_split(
    X, y, race_col, test_size=0.3, random_state=0, stratify=y)

model = LogisticRegression(max_iter=1000).fit(X_train, y_train)
print(f"Test accuracy: {model.score(X_test, y_test):.3f}")
```

```python
# False-positive and false-negative rates by group
y_pred = model.predict(X_test)
for g in sorted(race_test.unique()):
    mask = race_test == g
    tn, fp, fn, tp = confusion_matrix(y_test[mask], y_pred[mask]).ravel()
    fpr = fp / (fp + tn) if (fp + tn) else float("nan")
    fnr = fn / (fn + tp) if (fn + tp) else float("nan")
    print(f"Group {g}: n={mask.sum()}  FPR={fpr:.3f}  FNR={fnr:.3f}")
```

**Step 5: Interpret critically.**

We trained a model on synthetic data, deliberately constructed so that group A has higher income and credit scores on average than group B (a structural gap encoded into the simulation). The model never *saw* race as a feature, but it learned to approve at very different rates for the two groups *because the features it did see encoded the gap*.

This is the same pattern we saw with COMPAS in Week 3. It is a structural finding: when training data carries the imprint of a pre-existing inequity, race-blindness in features does not produce race-neutrality in outcomes. Any model on such data inherits the gap.

Note what *is not* claimed by this finding: it is **not** that algorithms are uniquely bad. It is that algorithms reproduce the patterns of the systems they're trained on. Fixing them well requires fixing those systems — or refusing to deploy automated decisions in domains where the underlying inequity has not been addressed.

**Step 6: Visualize for the digital essay.**

```python
# A clean, essay-ready figure
fig, ax = plt.subplots(figsize=(7, 4.5))
rates = []
for g in sorted(race_test.unique()):
    mask = race_test == g
    tn, fp, fn, tp = confusion_matrix(y_test[mask], y_pred[mask]).ravel()
    rates.append({"group": f"Group {g}",
                  "FPR": fp / (fp + tn) if (fp + tn) else 0,
                  "FNR": fn / (fn + tp) if (fn + tp) else 0})
rates_df = pd.DataFrame(rates).set_index("group")
rates_df.plot.bar(ax=ax, color=["#cc4444", "#4477aa"])
ax.set_ylabel("Error rate")
ax.set_title("Loan-approval model: error rates by group\n(model never used race as a feature)")
ax.legend(loc="upper right")
ax.text(0.5, -0.30,
        "Source: synthetic teaching dataset, DIGHUM 162 (n=200, structural gap encoded by design).",
        transform=ax.transAxes, ha="center", fontsize=8, style="italic")
plt.tight_layout()
plt.show()
```

💡 **Tip**: A polished figure for an essay needs a clear title, labeled axes, a source note, and consistent colors. Compare this final plot to the bar chart in Step 3 — both show the data; only one is essay-ready.

# 6. A Data Transparency Template for Your Project

Every dataset in your final project should come with a *data transparency note* in your digital essay. Use these prompts:

1. **Provenance.** Who collected this data, and for what purpose? When? In what jurisdiction? Was the collection consensual, mandatory, or covert?
2. **Inclusion.** Who is included in this dataset? At what rate? With what level of granularity?
3. **Exclusion.** Who is *not* included? Why? Are there populations the collection mechanism systematically misses (undocumented people, unbanked people, people without smartphones, people without housing)?
4. **Categories.** What categories does the dataset use, and what histories do those categories carry? (e.g., racial categories from the US Census, gender as binary, "household" as a social unit, etc.)
5. **Measurement.** What does each column actually *measure*, and what does that measurement *not* capture? (e.g., `priors_count` measures police contact, not crime.)
6. **Power.** Who benefits from the patterns this data surfaces? Who could be harmed by their disclosure?
7. **Reading.** What would the readings of this course (Hoffmann, O'Neil, Noble, Caliskan, D'Ignazio & Klein, Gil et al., Green) say about this dataset?

# 7. Digital Essay Structure Guide

A recommended outline for your individual digital essay:

1. **Introduction** — Your research question. Why it matters. What's at stake.
2. **Dataset** — What it is, where it comes from, who made it. Include your data transparency note.
3. **Method** — What you did, in beginner-friendly terms. A reader who's never written Python should be able to follow your description.
4. **Findings** — The patterns you uncovered. At least one figure with a source caption.
5. **Critical discussion** — What your findings mean *in light of structural inequity*. What they *don't* show. Where your method's blind spots are. Connect to at least three of the course readings.
6. **Conclusion** — Limits, future directions, recommendations (technical, policy, organizational, or refusal-to-deploy).
7. **References** — MLA format, per the syllabus.

⚠️ **Warning**: A common failure mode is essays that report patterns without interpreting them, or interpretations that don't engage with the readings. The course's whole premise is that the technical and the critical are inseparable. Your essay should embody that — not separate them into two unrelated halves.

# 8. Peer Review Checklist

When you read a classmate's project proposal, run through these:

- **Research question.** Is it specific? Tied to a dataset? Critical, or only descriptive?
- **Dataset choice.** Is the dataset accessible to them? Realistic in scope? Have they thought about its provenance?
- **Method.** Is the technique appropriate to the question? Do they have the skills to do it in two weeks?
- **Critical depth.** Are they engaging with at least three readings? Do those engagements feel substantive, or are they cited but not used?
- **Stakes.** Is it clear *who would care* about the result, and *why*? 
- **Risks.** Could the analysis cause harm if read by the wrong audience? Have they thought about that?
- **What's missing.** What did they leave out that they should consider adding? What's already too much that they should cut?

💡 **Tip**: Peer review is most useful when it's specific. "Tighten the introduction" is less useful than "the second paragraph of the introduction implies that the COMPAS dataset includes Hispanic defendants — but the analysis only includes Black and White defendants. Either expand the analysis or flag the limitation in the intro."

# 9. Common Pitfalls

Things to actively guard against:

- ⚠️ **Treating predictive accuracy as the goal.** O'Neil's critique is that high accuracy on a flawed task (predict re-arrest) just makes the flaw scale better. Always ask: *is this even the right question to be predicting?*
- ⚠️ **Ignoring who's missing from your dataset.** Erasure is data, too. A dataset of "all US households" misses unhoused people. A dataset of "all Twitter users in 2018" misses the global majority. Make absences visible.
- ⚠️ **Naturalizing constructed categories.** Race, gender, household, neighborhood, citizenship — all administrative and historical constructs. Use them when you must, but never write as though they were natural kinds.
- ⚠️ **Letting visualizations make claims your data can't support.** A choropleth at the state level can't show within-state variation. A correlation isn't a cause. A model's accuracy isn't a measure of fairness. Match the strength of your claim to the strength of your evidence.
- ⚠️ **Conflating *the model* with *the world*.** A word embedding's biases tell you about the corpus. A recidivism model's predictions tell you about prior arrests. Always say what the analysis is actually measuring.
- ⚠️ **Pretending you don't have a position.** D'Ignazio and Klein remind us: there is no view from nowhere. Acknowledging your own position, interests, and limits in your essay is intellectually serious, not a weakness.
